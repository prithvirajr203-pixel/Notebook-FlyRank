## 4. Claim rewrite

*Take your own boldest sentence and rewrite it in safe language: observed, measured, directional, decision-support.*

```python
# ML-09 — Section 4: Claim Rewrite

original_claim = (
    "The Random Forest model can accurately identify pages that are "
    "likely to decline and improve page-prioritization decisions."
)

safe_claim = (
    "In my evaluation, the Random Forest model showed measured "
    "performance in identifying pages associated with future "
    "search-performance decline. The result is directional evidence "
    "that the model may support page-prioritization decisions, but "
    "it does not establish that the model will generalize to all "
    "clients or future datasets. Further testing with unseen clients "
    "and future time periods would provide stronger evidence of "
    "generalization."
)

print("Original claim:")
print(original_claim)

print("\nSafe rewritten claim:")
print(safe_claim)

print("\nClaim language check:")
keywords = [
    "measured",
    "directional",
    "support",
    "does not establish",
    "future datasets"
]

for word in keywords:
    print(f"{word}: {'Present' if word.lower() in safe_claim.lower() else 'Missing'}")
```
