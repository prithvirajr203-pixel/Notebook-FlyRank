## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python
from IPython.display import Markdown, display

display(Markdown("""
# Target or Proxy

**Target:** Ranking Score

The model predicts a ranking score for each webpage, representing its expected SEO performance.

Since future search rankings are not directly available, a proxy target called **ranking_score** is used.

This label is created using existing SEO indicators such as content quality, keyword relevance, page performance, and engagement metrics. It serves as a practical approximation of future ranking performance.
"""))
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
from IPython.display import Markdown, display

display(Markdown("""
# Success Metric

**Success Metric:** Mean Absolute Error (MAE)

The model's performance will be evaluated using Mean Absolute Error (MAE), which measures the average difference between the predicted ranking score and the actual ranking score.

A lower MAE indicates better model performance.

For this project, an MAE of less than **5 points** would be considered a good result, meaning the predicted ranking scores are close to the expected values.
"""))
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
from IPython.display import Markdown, display

display(Markdown("""
# Why ML Beats a Fixed Rule Here

A fixed rule cannot accurately rank webpages because SEO performance depends on many factors working together.

For example:
- Content quality
- Keyword relevance
- User engagement
- Click-through rate (CTR)
- Page authority
- Search intent

These factors interact in complex ways that cannot be captured with simple if-else statements.

Machine Learning can learn these patterns from historical data and produce better ranking predictions, helping prioritize which content should be improved first.
"""))
```
