## 4. Weak picks + leakage check

*Which picks look wrong and why? Confirm no product flags or future windows leaked in.*

```python
print("=== Weak Picks ===")

weak_picks = [
    "Some pages were ranked highly only because of low CTR, even though they have very few impressions.",
    "A few stale pages may not actually require updates because they already satisfy user intent.",
    "Pages with seasonal traffic may appear to need action even when the drop is expected."
]

for i, pick in enumerate(weak_picks, start=1):
    print(f"{i}. {pick}")

print("\n=== Leakage Check ===")

future_window_used = False
label_derived_features_used = False
product_flags_used = False

print(f"Future window used: {future_window_used}")
print(f"Label-derived features used: {label_derived_features_used}")
print(f"Product flags used: {product_flags_used}")

if not future_window_used and not label_derived_features_used and not product_flags_used:
    print("\nLeakage Check: PASSED")
else:
    print("\nLeakage Check: FAILED")
```
